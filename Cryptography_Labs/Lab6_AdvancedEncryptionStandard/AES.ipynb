# AES Encryption and Decryption
AES example using a fixed message and key.

```python
from Cryptodome.Cipher import AES
from Cryptodome.Util.Padding import pad, unpad
from base64 import b64encode

message = b"Hello, this is my AES message"
key = b"thisisasecretkey"
mode = AES.MODE_ECB

cipher = AES.new(key, mode)
ciphertext = cipher.encrypt(pad(message, AES.block_size))
print("Encrypted:", b64encode(ciphertext).decode())

plain = unpad(AES.new(key, mode).decrypt(ciphertext), AES.block_size)
print("Decrypted:", plain.decode())
```
