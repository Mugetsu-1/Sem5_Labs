# AES CBC Mode
Encrypt and decrypt a message using AES in CBC mode.

```python
from Cryptodome.Cipher import AES
from Cryptodome.Util.Padding import pad, unpad
from base64 import b64encode

message = b"Hello, this is my CBC message"
key = b"thisisasecretkey"
iv = b"thisisaninitvect"
mode = AES.MODE_CBC

cipher = AES.new(key, mode, iv)
ciphertext = cipher.encrypt(pad(message, AES.block_size))
print("Encrypted:", b64encode(iv + ciphertext).decode())

plain = unpad(AES.new(key, mode, iv).decrypt(ciphertext), AES.block_size)
print("Decrypted:", plain.decode())
```
